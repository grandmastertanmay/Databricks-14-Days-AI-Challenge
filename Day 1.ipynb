## 🔥 Part 1: PySpark Warm-up
Before processing the main dataset, verifying the compute cluster with simple DataFrame operations.

```python
# Creating a dummy dataset to test Spark operations
data_products = [("iPhone", 999), ("Samsung", 799), ("MacBook", 1299)]
df_products = spark.createDataFrame(data_products, ["product", "price"])

print("--- Product Inventory ---")
df_products.show()

print("--- High Value Items (Filter > $1000) ---")
df_products.filter(col("price") > 1000).show()
```

## 📦 Part 2: Loading the eCommerce Dataset
The data has been pre-downloaded to Databricks Volumes. We will now read the raw CSV files into Spark DataFrames.

```python
# Define file paths (Pre-configured in Setup phase)
oct_path = "/Volumes/workspace/ecommerce/ecommerce_data/2019-Oct.csv"
nov_path = "/Volumes/workspace/ecommerce/ecommerce_data/2019-Nov.csv"

try:
    # Read CSVs with inferred schema
    print(f"⏳ Loading data from: {oct_path}...")
    df_oct = spark.read.option("header", "true").option("inferSchema", "true").csv(oct_path)
    
    print(f"⏳ Loading data from: {nov_path}...")
    df_nov = spark.read.option("header", "true").option("inferSchema", "true").csv(nov_path)
    
    print("✅ Data Ingestion Complete.")
    
except Exception as e:
    print(f"❌ Error loading data. Ensure Volumes are mounted correctly.\nDetails: {e}")
```

```python
# Count check to verify data volume
print(f"📊 October 2019 Events: {df_oct.count():,}")
print(f"📊 November 2019 Events: {df_nov.count():,}")

print("\n" + "="*50)
print("📌 Data Schema (October)")
print("="*50)
df_oct.printSchema()

print("\n" + "="*50)
print("📌 Sample Rows")
print("="*50)
display(df_oct.limit(5))
```
